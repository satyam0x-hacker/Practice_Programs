```python
# WAP to add a key and value to a dictionary
Dict={1:10,2:20,3:30}
Dict[4]=40
print(Dict)
```

```python
# WAP to cancatenate two dictionaries into a single dictionaries
D1={"A":1,"B":3,"C":5,"D":7}
D2={"E":2,"F":4,"G":6,"H":8}
New_Dict={}
New_Dict.update(D1)
New_Dict.update(D2)
print(New_Dict)
```

```python
# WAP to check whether a given key exists in the dictionary
Dict={"a":10,"b":20,"c":30,"d":40}
Key=input("Enter a key")
if Key in Dict:
    print("Yes the key exists in dictionary")
else:
    print("No such key found!")
    
```

```python
# WAP to print the keys alone, values alone and KEYS & VALUES alone from a dictionary
D={1:"Amit",2:"Rohan",3:"Mohan",4:"Sohan"}
for keys in D:
    print(keys)
for values in D.values():
    print(values)
for key,values in D.items():
    print(key,values)
    
```

```python
# WAP to prepare a dictionary where the keys are between 1 to 15 both included and their values are their respective squares
my_dict={}
for i in range(1,16):
    my_dict[i]=i**2
print(my_dict)
```

```python
# WAP to print the sum of all the values of a dictionary.
Dict={"A":10,"B":20,"C":30,"D":40}
total=0
for values in Dict.values():
    total+=values
print(total)
```

```python
# Converting a list into dictionary
L1=[1,3,5,7]
L2=[2,4,6,8]
Dictionary=dict(zip(L1,L2))
print(Dictionary)
```

```python
# Convert a dictionary into a list
Dict={1:10,2:20,3:30,4:40}
Keys=list(Dict.keys())
Values=list(Dict.values())
print("List of keys are:",Keys)
print("List of values are:",Values)
```

```python
# WAP to find the mean of the values of dictionary
Dict={"A":1,"B":2,"C":3,"D":4}
total=sum(Dict.values())
mean=total/len(Dict)
print(mean)
```

```python
# WAP to sort the elements of a dictionary
Dict={1:10,2:20,3:30,4:40}
Sorted_dict=sorted(Dict)
print(Sorted_dict)
```

```python
# WAP to create a dictionary of marks of students and display the highest marks.
Dict={"Satyam":100,"Pranit":97,"Rohit":83,"Mohit":20}
Highest_Marks=max(Dict.values())
print("Highest marks score are:",Highest_Marks)
Topper=max(Dict,key=Dict.get)
print(Topper)
```

```python
# WAP to count the frequency of each element in a list using dictionary
N1 = [10, 20, 10, 30, 20, 10, 40, 30] 
frequency = {} 
for i in N1:
    if i in frequency:
        frequency[i] = frequency[i] + 1 
else:
 frequency[i] = 1 
print("Frequency of elements:", frequency)

    


```
