```python
# 9. WAP to accept 20 values from the user and enter them in a list.
  # (a) Count similar elements of a list and print their index
  # (b) Count even and odd elements of a list
  # (c) Count positive and negative value of list
L=[int(input("Enter the values:")) for i in range(20)]

for i in range(len(L)):
    if L.count(L[i])>1:
        print(L[i], "It's index is:",i)
        
even=0
odd=0
for i in L:
    if i%2==0:
        even+=1
    else:
        odd+=1
print("Number of even elements are:",even)
print("Number of odd elements are:",odd)

positive=0
negative=0
for i in L:
    if i>0:
        positive+=1
    elif i<0:
        negative+=1
print("Number of positive elements are:",positive)
print("Numbre of negative elements are:",negative)
    
```

```python
# 10. Accept 10 values from the user and save it in a list.
  # (a) Sort list in ascending order using sorted() function and display sorted list.
  # (b) Sort list in descending order using sorted() function.
  # (c) display lenth of list

L=[int(input("Enter the element:")) for i in range(10)]
Ascending_L=sorted(L)
print(Ascending_L)

Descending_L=sorted(L,reverse=True)
print(Descending_L)

n=len(L)
print(n)
 ## Sorted() function defaultly returns the list in ascending order while using the line Sorted(L,reverse=True) can give descending order.
```

```python
# 17. WAP to merge two list as a dictionary one list as key and one list as it's items.
## so in order to merge two lists into a single dictionary format we use the ZIP() function in order to zip them into a single dictionary.
## we use Dict() function along with zip function
Key=[1,2,3,4,5]
value=["Eena","Meena","Dika","Dhaniya","Pudina","Achaar"]
Dictionary=dict(zip(Key,value))
print(Dictionary)
```
