```python
# To build a calculator to perform addition, subtraction, multiplication and division
n1=float(input("Enter the first number:"))
n2=float(input("Enter the second number:"))
Operator=input("Enter a operator (+,-,*,/):")
if Operator== "+":
    print("The sum of these numbers is:",n1+n2)
elif Operator== "-":
    print("The difference of these numbers is:",n1-n2)
elif Operator== "*":
    print("The multiplication is:",n1*n2)
elif Operator== "/":
    if n2!=0:
        print("The division is:",n1/n2)
    else:
        print("Zero Division ERROR!")
else:
    print("Invalid operator, GO GET A LIFE")

    
```

```python
# WAP to print the sum of all the digits of a number
n=int(input("Enter the two digit number:"))
SUM=0
while n>0:
    digit=n%10
    SUM=SUM + digit
    n=n//10

print(SUM)
```

```python
# To find the factorial of a given number
n=int(input("Enter the number:"))
factorial=1
for i in range(1,n+1):
    factorial=factorial*i
print(factorial)
```

```python
# To check whether the given number is prime or not
n=int(input("Enter a number:"))
count=0
for i in range(1,n+1):
    if n%i==0:
        count=count+1


if count== 2:
    print("Number is a prime number :)")
else:
    print("Number is not a prime number")
    

```

```python
# WAP to print the prime numbers between 10 to 99
for n in range(10,100):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    if count==2:
        print(n)
```

```python
# WAP to reverse a given number
n=int(input("Enter a number:"))
reverse=0
while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10

print("Reverse number is:",reverse)
```

```python
# WAP to check whether a given year is a leap year or not
## KEY THINGS TO KNOW BEFORE DOING THIS PROGRAM : 
## A year is a leap year only if: (i) Year is divisible by 400 (ii) Year is divisible by 4 but not divisible by 100
Year=int(input("Enter a year"))
if Year % 400 == 0:
    print("Yes this is a leap year")
elif Year % 100 == 0:
    print("No this is not a leap year")
elif Year % 4 == 0:
    print("Yes it is a leap year")
else:
    print("Not a leap year")
    
```
