```python
## TUPLES PROGRAMS -

# WAP to print the 4th element from first and 4th element from the last of a tuple.
Numbers=(1,2,3,4,5,6,)
print("4th number from the first is:",Numbers[3])
print("4th number from the last is:",Numbers[-4])
```

```python
# WAP to check whether an element exists in the tuple or not.
Tuple=(1,2,3,4,5)
N=int(input("Enter the number to searched:"))
if N in Tuple:
    print("Element exists in Tuple")
else:
    print("No such element found!")
```

```python
# WAP to convert a list into a tuple.
List=[10,20,30,40,50,60]
Tuple=tuple(List)
print(Tuple)

```

```python
# WAP to find the index of an item in tuple.
Tuple=(10,50,80,70,40,60,90)
element=int(input("Enter the element whose index is to be found:"))
if element in Tuple:
    print("Index of element is:",Tuple.index(element))
else:
    print("Something has gone wrong!")
    
```

```python
# WAP to replace the last element in a list with 100.
List=[(10,20,30),(40,50,60),(70,80,90)]
New_List=[]
for i in List:
    Tuple=i[:-1] + (100,)
    New_List.append(Tuple)
print(New_List)
    
```

```python
# WAP to find the maximum and minimum of elements inside a tuple.
Tuple=(1,2,3,4,5)
print("Maximum of these is:",max(Tuple))
print("Minimum of these is:",min(Tuple))
```

```python
# WAP to swap the first and last element of a tuple.
Tuple=(5,2,3,4,1)
New_Tuple=(Tuple[-1],) + (Tuple[1:-1],) + (Tuple[0],)
print(New_Tuple)
```
